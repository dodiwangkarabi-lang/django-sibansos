```python
import pandas as pd
from openpyxl import Workbook
from openpyxl.styles import Alignment, Font

# contoh data
df = pd.DataFrame({
    "Nama": ["A", "B", "C"],
    "Nilai": [80, 90, 85]
})

# buat workbook kosong
wb = Workbook()
ws = wb.active

# =========================
# 1. KOP SURAT
# =========================
ws.merge_cells("A1:D1")
ws["A1"] = "DINAS PENDIDIKAN"
ws["A1"].font = Font(size=14, bold=True)
ws["A1"].alignment = Alignment(horizontal="center")

ws.merge_cells("A2:D2")
ws["A2"] = "LAPORAN NILAI MAHASISWA"
ws["A2"].alignment = Alignment(horizontal="center")

# =========================
# 2. KETERANGAN
# =========================
ws["A4"] = "Keterangan: Laporan nilai semester ganjil"
ws["A4"].font = Font(italic=True)

# =========================
# 3. TABEL (DATAFRAME)
# =========================
start_row = 6

for col_idx, col_name in enumerate(df.columns, 1):
    ws.cell(row=start_row, column=col_idx, value=col_name).font = Font(bold=True)

for row_idx, row in enumerate(df.itertuples(index=False), start=start_row + 1):
    for col_idx, value in enumerate(row, 1):
        ws.cell(row=row_idx, column=col_idx, value=value)

# =========================
# 4. TANDA TANGAN (POJOK KIRI BAWAH)
# =========================
tt_row = start_row + len(df) + 5

ws["A{}".format(tt_row)] = "Makassar, 18 Juni 2026"
ws["A{}".format(tt_row + 1)] = "Mengetahui,"
ws["A{}".format(tt_row + 4)] = "_____________________"
ws["A{}".format(tt_row + 5)] = "Nama Pejabat"

# simpan file
wb.save("laporan.xlsx")
```
